```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer # Tokenizer is deprecated instead use text_vectorization
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentenses = [
    'I love machine learning',
    'I love machine learning and natural language processing',
    'Do you love machine learning?', # Notice that machine learning is repeated in the first three sentences, one with a question mark at the end. This will affect the tokenization process.
    'Natural language processing is fun',
    'Tokenization is important for NLP',
    'NLP is the future of AI',
    'Learning to tokenize text is essential'
]
```

```python
tokernizer = Tokenizer(num_words=100, oov_token='<OOV>')
tokernizer.fit_on_texts(sentenses)
```

```python
word_index = tokernizer.word_index
print(word_index)
```

```python
for word, index in word_index.items():
    print(f"{index}: {word}")
```
