### PART 1: Word Level Tokenization using TextVectorization in Keras

```python
# Define the TextVectorization layer

# Create a TextVectorization layer that:
# - Converts text to lowercase and strips punctuation
# - Splits text into tokens by whitespace
# - Outputs integer indices for each token
# - Pads/truncates sequences to a fixed length of 12
word_vectorizer = tf.keras.layers.TextVectorization(
    standardize='lower_and_strip_punctuation',
    split='whitespace',
    output_mode='int',
    output_sequence_length=12
)
```

```python
# Adapt the vectorizer to the corpus
# 'corpus' should be a list of sentences or text samples
# This step builds the vocabulary based on the corpus
word_vectorizer.adapt(corpus)
```

```python
# Inspect the vocabulary
# Retrieve the vocabulary learned from the corpus
word_vocab = word_vectorizer.get_vocabulary()

# Print vocabulary size and the actual vocabulary list
print(f"\nVocabulary size: {len(word_vocab)}")
print(f"Vocabulary: {word_vocab}")
```

```python
# Encode the corpus into integer sequences
# Transform the corpus into sequences of integers
# Each word is replaced by its index in the vocabulary
word_encoded = word_vectorizer(corpus)

print("\nEncoded sequences (sentences X max_len):")
print(word_encoded.numpy())
```

```python
# Decode an encoded sentence back to words
# Create a mapping from index to word
idx2word = {i: w for i, w in enumerate(word_vocab)}
idx2word
```

```python
# Decode the first sentence (index 0) by mapping indices back to words
# Skip padding tokens (index 0)
print("\nDecoded sentence 0:")
print(" ".join([idx2word[i] for i in word_encoded[0].numpy() if i != 0]))
```

```python
# Test handling of unseen (OOV) words
# Provide a sentence with words not in the training corpus
unseen = ["Unseen vocabulary handling works via unknown tokens"]

print(f"\nOOV Test — '{unseen[0]}'")
print("Encoded:", word_vectorizer(unseen).numpy())
print("→ Token index 1 = [UNK]")  # Index 1 is reserved for unknown words
```

### Word Level Tokenization V2

```python
max_vocab_size = 20

word_vectorizer = tf.keras.layers.TextVectorization(
    max_tokens=max_vocab_size,
    output_mode="int",
    output_sequence_length=6  # pad/truncate
)

# Adapt builds vocabulary
word_vectorizer.adapt(corpus)

# View vocabulary
word_vocab = word_vectorizer.get_vocabulary()
print("Word Vocabulary:")
print(word_vocab)
print("Vocabulary Size:", len(word_vocab))
```

### Encode Sentences

```python
word_encoded = word_vectorizer(corpus)
print("WOrd-Level Encoded Sequences: ")
print(word_encoded.numpy())
```
